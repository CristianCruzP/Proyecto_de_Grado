# 05 Evaluation


Evaluación cronológica con baseline ingenuo y limitaciones actuales.


```python
import numpy as np
import pandas as pd
from sklearn.metrics import mean_absolute_error, mean_squared_error
df = pd.read_csv('data/gold/dataset_lstm_mensual.csv')
valid = df['drought_label'].dropna().to_numpy()
if len(valid) > 1:
    y_true = valid[1:]
    y_pred = valid[:-1]
    print({'baseline_mae': float(mean_absolute_error(y_true,y_pred)), 'baseline_rmse': float(np.sqrt(mean_squared_error(y_true,y_pred)))})
else:
    print('Cobertura insuficiente para baseline')
```
